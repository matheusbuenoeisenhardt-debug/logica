2)

remover da fila (FIFO)

mostrar cliente atendido

salvar na pilha histórico

```python
fila = []
histórico = []

while True:
  escolha = input('insira 1 para adicionar um cliente, 2 para atender cliente, 3 para mostrar fila, 4 para mostrar histórico, 5 para buscar cliente ou 6 para sair: ')

  if escolha == '1':
    nome = input("digite seu nome: ")
    idade = input("digite sua idade: ")
    cliente = {'nome': nome, 'idade': idade}
    print(cliente)

  elif escolha == '2':
    print('atender o cliente')
    fila.pop(0)
    histórico.append(cliente)

  elif escolha == '3':
    print('mostrar fila')

  elif escolha == '4':
    print('mostrar histórico')

  elif escolha == '5':
    print('buscar cliente')

  elif escolha == '6':
    print('sistema encerrando')
    break
```

3)

exibir todos os clientes na ordem de atendimento

```python
fila = []
histórico = []
print('digite 1 para adicionar um cliente')
print('digite 2 para atender um cliente')
print('digite 3 para mostrar a fila')
print('digite 4 para mostrar o histórico')
print('digite 5 para buscar um cliente')
print('digite 6 para sair')

while True:
  escolha = input('qual opção deseja escolher?: ')

  if escolha == '1':
    nome = input("digite seu nome: ")
    idade = input("digite sua idade: ")
    cliente = {'nome': nome, 'idade': idade}
    fila.append(cliente)
    print('cliente adicionado a fila')

  elif escolha == '2':
    if not fila:
      print('fila vazia')
    else:
      print('atender o cliente')
      cliente = fila.pop(0)
      histórico.append(cliente)
      print(histórico)

  elif escolha == '3':
    if not fila:
      print('fila vazia')
    else:
      print('mostrar fila')
      for cliente in fila:
        print(cliente)

  elif escolha == '4':
    print('mostrar histórico')

  elif escolha == '5':
    print('buscar cliente')

  elif escolha == '6':
    print('sistema encerrando')
    break
```

4)

mostrar os clientes já atendidos

ordem: último atendido no começo (LIFO)

```python
fila = []
histórico = []
print('digite 1 para adicionar um cliente')
print('digite 2 para atender um cliente')
print('digite 3 para mostrar a fila')
print('digite 4 para mostrar o histórico')
print('digite 5 para buscar um cliente')
print('digite 6 para sair')

while True:

  escolha = input('\nqual opção deseja escolher?: ')

  if escolha == '1':
    nome = input("\ndigite seu nome: ")
    idade = input("digite sua idade: ")
    cliente = {'nome': nome, 'idade': idade}
    fila.append(cliente)
    print('cliente adicionado a fila')

  elif escolha == '2':
    if not fila:
      print('\nfila vazia')
    else:
      print('\natender o cliente')
      cliente = fila.pop(0)
      histórico.append(cliente)
      print(f'cliente {cliente} adicionado ao histórico')

  elif escolha == '3':
    if not fila:
      print('\nfila vazia')
    else:
      print('\nfila de atendimentos: ')
      for cliente in fila:
        print(cliente['nome'], cliente['idade'])

  elif escolha == '4':
    if not histórico:
      print('\nhistórico vazio')
    else:
      print('\nhistórico de atendimentos: ')
      for cliente in reversed(histórico):
        print(cliente['nome'], cliente['idade'])

  elif escolha == '5':
    print('\nbuscar cliente')

  elif escolha == '6':
    print('\nSistema encerrando')
    break
```

```python
fila = []
histórico = []
print('digite 1 para adicionar um cliente')
print('digite 2 para atender um cliente')
print('digite 3 para mostrar a fila')
print('digite 4 para mostrar o histórico')
print('digite 5 para buscar um cliente')
print('digite 6 para sair')

while True:

  escolha = input('\nqual opção deseja escolher?: ')

  if escolha == '1':
    nome = input("\ndigite seu nome: ")
    idade = input("digite sua idade: ")
    cliente = {'nome': nome, 'idade': idade}
    fila.append(cliente)
    print('cliente adicionado a fila')

  elif escolha == '2':
    if not fila:
      print('\nfila vazia')
    else:
      print('\natender o cliente')
      cliente = fila.pop(0)
      histórico.append(cliente)
      print(f'cliente {cliente} adicionado ao histórico')

  elif escolha == '3':
    if not fila:
      print('\nfila vazia')
    else:
      print('\nfila de atendimentos: ')
      for cliente in fila:
        print(cliente['nome'], cliente['idade'])

  elif escolha == '4':
    if not histórico:
      print('\nhistórico vazio')
    else:
      print('\nhistórico de atendimentos: ')
      for cliente in reversed(histórico):
        print(cliente['nome'], cliente['idade'])

  elif escolha == '5':
    cliente_buscado = input('\nqual cliente deseja procurar: ')
    for cliente in fila:
      if cliente['nome'] == cliente_buscado:
        print(f'{cliente['nome']} está na fila e tem {cliente['idade']} anos:')
    for cliente in histórico:
      if cliente['nome'] == cliente_buscado:
        print(f'{cliente['nome']} está no histórico e tem {cliente['idade']} anos')

  elif escolha == '6':
    print('\nSistema encerrando')
    break
```
