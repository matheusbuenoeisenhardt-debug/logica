calcule uma média de nota

```python
notas = [7.5, 8.0, 6.5, 9.0, 8.5]
soma = sum(notas)
media = soma / len(notas)

print(f"Notas digitadas: {notas}")
print(f"media: {media}")
```

```python
notas = [7.5, 8.0, 6.5, 9.0, 8.5]
soma = 0
x = 0

while x < 5:
  soma = soma + notas[x]
  x = x + 1

print(f'media = {soma / 5}')
```

digitando os números

```python
notas = []
soma = 0
x = 0

while x < 5:
  nota_digitada = float(input(f'insira a {x+1}ª nota: '))
  notas.append(nota_digitada)
  soma = soma + nota_digitada
  x = x + 1

media = soma / len(notas)

print(f'Notas digitadas: {notas}')
print(f'media = {media}')
```

slicing (delimita qual área da lista será exibida)

```python
lista = [1, 2, 3, 4, 5]
cópia = lista [0:5] # [1, 2, 3, 4, 5]
lista [:5] # [1, 2, 3, 4, 5]
lista [1:3] # [2, 3]
lista [1:4] # [2, 3, 4]
bolsonaro = lista [3:] # [4, 5]
lista [:3] # [1, 2, 3]
print(bolsonaro)
print(cópia)
print(lista)
```

sinal de negativo serve para desconsiderar o número de casas no final

digitando a lista com input

```python
lista = []
while True:
  entrada = input('digite um número inteiro para a lista: (digite 999 para sair): ')
  if entrada == '999':
    break
  numero = int(entrada)
  lista.append(numero)
print(f'a lista digitada foi: {lista}')
```
