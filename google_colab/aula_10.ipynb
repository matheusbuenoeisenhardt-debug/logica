crie um dicionário com nome do aluno como chave e nota como valor

.

permita cadastrar 3 alunos
mostre todos os alunos e suas notas

```python
alunos = {}

for i in range(2):
  nome = input('digite o nome do aluno: ')
  nota = float(input('digite a nota do aluno: '))      # O float serve para que possa ser enviado um número com casas decimais
  alunos[nome] = nota

print()
soma = 0
for nome, nota in alunos.items():                       # Cita os dois termos que vão ser usados no for
  print(f'Aluno: {nome} - Nota: {nota:.2f}')            # Utiliza as duas variaveis e ao mesmo tempo arredonda a nota
  soma = soma + nota                                    # Adiciona a nota para a soma final que será dividida pelos alunos
print()

media = soma / len(alunos)                              # Utiliza o len(alunos) para ver quantas respostas tiveram e assim saber o divisor

print(f'a media é {media:.2f}')
```

lista de palavras

```python
#cria um dicionario vazio
dicionario = {}

frase = input('digite uma frase: ')

#separando as palavras
lista_palavras = frase.split()

print(lista_palavras)

for palavra in lista_palavras:
  if palavra not in dicionario:
    dicionario[palavra] = 1
  else:
    dicionario[palavra] = dicionario[palavra] + 1

print(dicionario)
```
